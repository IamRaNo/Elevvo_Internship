```python
def hour_classification(hour):
    if 5 <= hour < 8:
        return 'early morning'
    elif 8 <= hour < 12:
        return 'morning'
    elif 12 <= hour < 16:
        return 'afternoon'
    elif 16 <= hour < 20:
        return 'evening'
    elif 20 <= hour < 24:
        return 'night'
    else:  # 0 <= hour < 5
        return 'midnight'
```

```python
data['purchase_time']=data['purchase_date'].dt.hour.apply(hour_classification)
```

```python
data['purchase_date']=pd.to_datetime(data['purchase_date'].dt.date)
```

```python
data['approve_time']=data['approve_date'].dt.hour.apply(hour_classification)
```

```python
data['approve_date']=pd.to_datetime(data['approve_date'].dt.date)
```

```python
data['delivered_carrier_time']=data['delivered_carrier_date'].dt.hour.apply(hour_classification)
```

```python
data['delivered_carrier_date']=pd.to_datetime(data['delivered_carrier_date'].dt.date)
```

```python
data['delivered_customer_time']=data['delivered_customer_date'].dt.hour.apply(hour_classification)
```

```python
data['delivered_customer_date']=pd.to_datetime(data['delivered_customer_date'].dt.date)
```

```python
def cancel_rate(x):
    return (x == 'canceled').sum() / x.nunique()


def safe_mode(x):
    m = x.mode()
    return m.iloc[0] if not m.empty else None

customer_data=data.groupby('customer_id').agg({
        'status': cancel_rate,
        'purchase_date': 'min',
        'customer_city': 'first',
        'customer_state': 'first',
        'customer_zip': 'first',
        'order_item_id': 'count',
        'price': ['sum', 'mean'],
        'freight_value': ['sum', 'mean'],
        'payment_type': safe_mode,
        'product_weight_g': 'median',
        'seller_id': safe_mode,
        'seller_city': safe_mode,
        'seller_state': safe_mode,
        'product_category': safe_mode,
        'purchase_time': safe_mode,
        'delivery_days': 'mean',
        'used_installments': lambda x: int((x > 0).any()),
        'used_sequential': lambda x: int((x > 0).any())
    }).reset_index()
```

```python
customer_data.columns = ['_'.join(col).strip('_') for col in customer_data.columns.values]
```
